```python
from pymongo import MongoClient
import numpy as np

def calculate_average_salary():
    client = MongoClient('mongodb://localhost:27017/')
    db = client['mongodump --uri mongodb+srv://ihrmaiml:<PASSWORD>@ast.7rrpg.mongodb.net/<DATABASE> ']
    collection = db['python_jobs']


    jobs = list(collection.find())


    salaries = [job['salary'] for job in jobs if 'salary' in job]

    if salaries:

        salaries_array = np.array(salaries)
        average_salary = np.mean(salaries_array)
        print(f"The average salary is: ${average_salary:.2f}")
    else:
        print("No salary data available.")

calculate_average_salary()
```
